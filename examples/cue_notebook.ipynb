# Manim Cue in a notebook
Cue runs the cells that the scene needs.

```python
from manim import *
```

```python
# A slow setup cell that the scene does not need: Cue skips it.
raise RuntimeError('This cell must not run')
```

```python
RADIUS = 1.5
COLORS = [TEAL, YELLOW]
```

```python
%%manim -qm Orbit
class Orbit(Scene):
    def construct(self):
        circle = Circle(radius=RADIUS, color=COLORS[0])
        dot = Dot(color=COLORS[1]).move_to(circle.point_from_proportion(0))
        self.play(Create(circle), run_time=1)
        self.play(MoveAlongPath(dot, circle), run_time=1)
        self.wait(0.5)
```
